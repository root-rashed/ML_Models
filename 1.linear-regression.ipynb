```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)

import os
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
import kagglehub
```

```python
df = pd.read_csv("/kaggle/input/datasets/nabeelqureshitiii/student-performance-dataset/student_performance.csv")
```

```python
df.sample(5)
```

```python
df.info()
```

```python
X= df[['weekly_self_study_hours','attendance_percentage','class_participation']]
```

```python
y = df['total_score']
```

```python
X.shape
```

```python
y.shape
```

**Datset Split and processing**

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
X_train.sample(5)
```

```python
X_test.sample(5)
```

```python
y_train.sample(5)
```

```python
y_test.sample(5)
```

**Now Import Model**

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

**Giving Training Data to model**

```python
model.fit(X_train,y_train)
```

**Now prediction**

```python
y_pred = model.predict(X_test)
```

```python
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
```

**Now testing Accuracy**

```python
from sklearn.metrics import r2_score
r2 = r2_score(y_test, y_pred)
```

```python
print("R² Score:", r2)
print("R² Percentage:", r2 * 100, "%")
```

**Now Testing accuracy by given user Input**

```python
study_hours = float(input("Weekly self-study hours: "))
attendance = float(input("Attendance percentage: "))
participation = float(input("Class participation: "))

new_student = np.array([[study_hours, attendance, participation]])

prediction = model.predict(new_student)

print("Predicted Total Score:", prediction[0])
```
