**Now the Main Part of OrdinalEncoder**

```python
from sklearn.preprocessing import OrdinalEncoder
```

**OrdinalEncoder converts categorical input columns into numbers.**

**Poor     → 0
Average  → 1
Good     → 2**

**School → 0
UG     → 1
PG     → 2**

```python
oe = OrdinalEncoder(categories=[
        ['Poor','Average','Good'],
        ['School','UG','PG']
    ]
)
```

**fit()**

**Learn the categories from my training data."**

```python
oe.fit(X_train)
```

**transform()**

```python
X_train = oe.transform(X_train)
```
